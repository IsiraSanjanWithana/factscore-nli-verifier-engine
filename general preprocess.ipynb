```python
import pandas as pd
import numpy as np
```

## Load the labeled (annotated) data

```python
annotated_df = pd.read_json("./data/raw/dev_labeled_ChatGPT.jsonl", lines=True)
print(f"loaded bios: {annotated_df.shape}")
annotated_df.head()
```

```python
# strip the "Question: " prefix from the input column
annotated_df["input"] = annotated_df["input"].str.replace("Question: ", "", regex=False)
```

## Coverting the annotations lsit to single value column

```python
df = annotated_df.explode('annotations',ignore_index=True)
print(f"before exploding annotations: {df.shape}")
print(f"after exploding annotations: {df.shape}")
```

## Converting annotations column's JSON key values pair to colums and rows

```python
annotations = pd.json_normalize(df['annotations'])
print(f"annotations shape: {annotations.shape}")
print(f"annotations columns: {annotations.columns}")
```

## Concat the df and exracted annotations

```python
print(f"annotations shape: {annotations.shape}")
print(f"df shape: {df.shape}")
```

```python
df = pd.concat([df.drop(columns=["annotations"], inplace=False), annotations], axis=1)
print(f"df shape: {df.shape}")
print(f"df columns: {df.columns}")
```

```python
df[:1]
```

```python
print(f"pre shape: {df.shape}")

df = df.explode('human-atomic-facts',ignore_index=True)

print(f"post shape: {df.shape}")
```

```python
df[:1]
```

## Converting human-atomic-facts column's JSON key values pair to colums and rows

```python
human_facts = pd.json_normalize(df['human-atomic-facts'])
human_facts.rename(columns={'text': "atomic-fact"}, inplace=True)

print(f"post shape: {human_facts.shape}")
print(f"post columns: {human_facts.columns}")
human_facts.head(5)
```

## Concat the df and exracted human-atomic-facts

```python
print(f"post shape: {df.shape}")
print(f"post columns: {df.columns}")

df = df.reset_index(drop=True)
human_facts = human_facts.reset_index(drop=True)
df = pd.concat([df, human_facts], axis=1)

print(f"post shape: {df.shape}")
print(f"post columns: {df.columns}")

df.head(5)
```

## Removing the unnessary columns

```python
df.drop(columns=["output","is-relevant","model-atomic-facts","human-atomic-facts"], inplace=True)
df.columns
```

## Checking the duplicaed rows according tpoic and atomic-fact

```python
df[df.duplicated(["topic", "atomic-fact"], keep=False)].sort_values("atomic-fact")[["topic", "atomic-fact", "label"]]
```

```python
df['label'].value_counts()
```

## Replacing the IR label with NS

```python
df["label"] = df["label"].replace("IR", "NS")
df["label"].value_counts()
```

```python
annotated_df = df
print(f"annotated_df shape: {annotated_df.shape}")
annotated_df.head(10)
```

## Load the Wikipedia passages (BM25-retrieved evidence)

```python
passages_df = pd.read_json("./data/raw/passages_bm25_ChatGPT_humfacts.jsonl", lines=True)
print(f"shape: {passages_df.shape}")
passages_df.head(5)
```

## Extract the text and convert text into a list

```python
passages_df["passage_list"] = passages_df["passages"].apply(lambda ps: [p["text"] for p in ps])
print(f"passage_text shape: {passages_df['passage_list'].shape}")
passages_df["passage_list"].head(5)
```

```python
passages_df.head(10)
```

## Merge facts with their passages

```python
print(f"pre-merge annotated: {annotated_df.shape}")
print(f"pre-merge passages:  {passages_df.shape}")

df = pd.merge(
    annotated_df,
    passages_df,
    left_on=["topic", "atomic-fact"],
    right_on=["name", "sent"],
    how="left"
)

print(f"post-merged shape: {df.shape}")
print(f"post-merged columns: {df.columns}")
df.head(5)
```

## Drop the redundant coluns generated while mergeing

```python
df.drop(columns=["name", "sent","passages",], inplace=True)
```

## Checking the duplicaed rows according tpoic and atomic-fact

```python
mask = df.duplicated(subset=["topic", "atomic-fact"], keep=False)
df[mask].sort_values("atomic-fact")[["topic", "atomic-fact", "label"]]
```

```python
df.rename(columns={"passage_list": "reference_passages"}, inplace=True)
df.columns
```

## Checking Null value and Empty values

```python
print(f"Number of null values in each column:\n{df.isnull().sum()}")
print(f"Number of empty lists in reference_passages column:\n{df['reference_passages'].apply(lambda x: isinstance(x,list) and len(x)==0).sum()}")
print("rows with \ufffd:", df["reference_passages"].apply(lambda ps: any("\ufffd" in p for p in ps)).sum())
```

```python
df["reference_passages"] = df["reference_passages"].apply(
    lambda ps: [p.replace("\ufffd", "") for p in ps]
)
print("rows with \ufffd:", df["reference_passages"].apply(lambda ps: any("\ufffd" in p for p in ps)).sum())
```

```python
def preprocess_passages(evidence_passages, topic):
    count = 0
    senteces = ""
    for i in "".join(evidence_passages).split("</s><s>"):
        count+=1
        if count == 1:
            i = i.strip("<s> ")
            if i.startswith(topic+ " "):
                i = i[len(topic):].strip()
        else:
            i = " ".join(i.split(" ")[1::])
        if i == "":
            continue
        senteces += i + " "
    return senteces.strip()
```

```python
df["evidence_passages"] = df.apply(lambda row: preprocess_passages(row["evidence_passages"], row["topic"]), axis=1)
```
